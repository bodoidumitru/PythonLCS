```python
import pandas as pd

def find_lcs(text1, text2):
    n, m = len(text1), len(text2)
    dp = [[0] * (m + 1) for _ in range(n + 1)]

    for i in range(1, n + 1):
        for j in range(1, m + 1):
            if text1[i - 1] == text2[j - 1]:
                dp[i][j] = dp[i - 1][j - 1] + 1
            else:
                dp[i][j] = max(dp[i - 1][j], dp[i][j - 1])

    result = []
    i, j = n, m
    while i > 0 and j > 0:
        if text1[i - 1] == text2[j - 1]:
            result.append(text1[i - 1])
            i -= 1
            j -= 1
        elif dp[i - 1][j] >= dp[i][j - 1]:
            i -= 1
        else:
            j -= 1

    result.reverse()
    return "".join(result)

def analyze_csv(input_file, output_file):
    data = pd.read_csv(input_file, keep_default_na=False)
    required = {"Text1", "Text2"}
    if not required.issubset(data.columns):
        raise ValueError("CSV must contain columns named Text1 and Text2")

    results = []
    for _, row in data.iterrows():
        text1, text2 = str(row["Text1"]), str(row["Text2"])
        lcs = find_lcs(text1, text2)
        results.append({
            "Text1": text1,
            "Text2": text2,
            "LCS": lcs,
            "LCS_Length": len(lcs),
            "Text1_Length": len(text1),
            "Text2_Length": len(text2)
        })

    result_data = pd.DataFrame(results)
    result_data.to_csv(output_file, index=False, encoding="utf-8-sig")
    print(result_data[["LCS", "LCS_Length", "Text1_Length", "Text2_Length"]])
    print(f"\nSaved results to: {output_file}")

if __name__ == "__main__":
    input_file = input("Enter input CSV filename (e.g. 01_text_comparison.csv): ").strip()
    output_file = input("Enter output CSV filename (e.g. result.csv): ").strip()
    analyze_csv(input_file, output_file)
```

```python
### TESTING ####
def visualize_lcs(text1, text2):
    """
    Пошагово показывает, как заполняется таблица dp.
    """
    n = len(text1)
    m = len(text2)

    print(f"Сравниваем: '{text1}' и '{text2}'")
    print(f"Длина первой: {n}, длина второй: {m}\n")

    # Создаем таблицу
    dp = [[0] * (m + 1) for _ in range(n + 1)]

    # --- ШАГ 1: ПОКАЗЫВАЕМ ПУСТУЮ ТАБЛИЦУ ---
    print("ШАГ 1: Изначально таблица заполнена нулями.")
    print("Строки = буквы первой строки (+1 пустая), Столбцы = буквы второй строки (+1 пустой)\n")
    print_table(dp, text1, text2)

    # --- ШАГ 2: ЗАПОЛНЯЕМ И КОММЕНТИРУЕМ ---
    print("\nШАГ 2: Заполняем таблицу по правилам:\n")
    print("  - Если буквы совпали -> берем диагональ (верх-лево) + 1")
    print("  - Если не совпали -> берем max(сверху, слева)\n")

    for i in range(1, n + 1):
        for j in range(1, m + 1):
            char1 = text1[i - 1]
            char2 = text2[j - 1]

            if char1 == char2:
                dp[i][j] = dp[i - 1][j - 1] + 1
                reason = f"'{char1}' == '{char2}' -> диагональ ({dp[i-1][j-1]}) + 1 = {dp[i][j]}"
            else:
                top = dp[i - 1][j]
                left = dp[i][j - 1]
                dp[i][j] = max(top, left)
                reason = f"'{char1}' != '{char2}' -> max(сверху {top}, слева {left}) = {dp[i][j]}"

            print(f"  Ячейка [{i}][{j}] ({char1} vs {char2}): {reason}")

    # --- ШАГ 3: ПОКАЗЫВАЕМ ИТОГОВУЮ ТАБЛИЦУ ---
    print("\nШАГ 3: Итоговая таблица (числа = длина LCS для префиксов)\n")
    print_table(dp, text1, text2)

    print(f"\nОтвет (правый нижний угол): {dp[n][m]}")

    # --- ШАГ 4: ВОССТАНОВЛЕНИЕ ПУТИ ---
    print("\nШАГ 4: Идем обратно от правого нижнего угла, чтобы собрать саму строку.\n")

    result = []
    i, j = n, m
    path = []

    while i > 0 and j > 0:
        char1 = text1[i - 1]
        char2 = text2[j - 1]

        if char1 == char2:
            result.append(char1)
            path.append(f"  [{i}][{j}] '{char1}' == '{char2}' -> БЕРЕМ букву, идем влево-вверх")
            i -= 1
            j -= 1
        elif dp[i - 1][j] >= dp[i][j - 1]:
            path.append(f"  [{i}][{j}] '{char1}' != '{char2}' -> идем ВВЕРХ (сверху число больше)")
            i -= 1
        else:
            path.append(f"  [{i}][{j}] '{char1}' != '{char2}' -> идем ВЛЕВО (слева число больше)")
            j -= 1

    for step in path:
        print(step)

    result.reverse()
    final = "".join(result)
    print(f"\nСобранная строка: '{final}' (длина {len(final)})")
    return final


def print_table(dp, text1, text2):
    """
    Красиво печатает таблицу с заголовками.
    """
    # Заголовок столбцов
    header = "      " + "  ".join(["_"] + list(text2))
    print(header)

    for i, row in enumerate(dp):
        # Заголовок строки
        row_label = "_" if i == 0 else text1[i - 1]
        row_str = f"  {row_label} | "
        for val in row:
            row_str += f"{val:>2} "
        print(row_str)
    print()


# --- ПРИМЕР ЗАПУСКА ---
if __name__ == "__main__":
    visualize_lcs("ABC", "AC")
```
